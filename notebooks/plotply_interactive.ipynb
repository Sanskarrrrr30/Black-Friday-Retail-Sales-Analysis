```python
# Type conversions

df['original_price'] = df['original_price'].astype('float64')
df['discount_pct'] = df['discount_pct'].astype('float64')
df['final_price'] = df['final_price'].astype('float64')
df['quantity'] = df['quantity'].astype('int64')
df['customer_id'] = df['customer_id'].astype('str')
df['transaction_id'] = df['transaction_id'].astype('str')
df['product_id'] = df['product_id'].astype('str')
df["purchase_date"] = pd.to_datetime(df["purchase_date"])

df.dtypes
```

```python
black_friday_df = df[df["is_black_friday"] == True]

normal_df = df[df["is_black_friday"] == False]
```

```python
gender_normal = (normal_df.groupby("gender")["purchase_amount"].sum().reset_index())
gender_normal["sale_type"] = "Normal"

gender_bf = (black_friday_df.groupby("gender")["purchase_amount"].sum().reset_index())
gender_bf["sale_type"] = "Black Friday"

gender_compare = pd.concat([gender_normal, gender_bf])
```

```python
import plotly.express as px

fig_gender_donut = px.pie(gender_bf,names="gender", values="purchase_amount", hole=0.5, title="Black Friday Gender Contribution")

fig_gender_donut.show()
```

```python
age_normal = (normal_df.groupby("age_group")["purchase_amount"].sum().reset_index())
age_normal["sale_type"] = "Normal"

age_bf = (black_friday_df.groupby("age_group")["purchase_amount"].sum().reset_index())
age_bf["sale_type"] = "Black Friday"

age_compare = pd.concat([age_normal, age_bf])
```

```python
fig_age = px.bar(age_compare, x="age_group", y="purchase_amount", color="sale_type", barmode="group", title="Age Group Spending Comparison",labels={"purchase_amount": "Revenue", "age_group": "Age Group"})

fig_age.show()
```

```python
category_normal = (normal_df.groupby("product_category")["purchase_amount"].sum().reset_index())
category_normal["sale_type"] = "Normal"


category_bf = (black_friday_df.groupby("product_category")["purchase_amount"].sum().reset_index())
category_bf["sale_type"] = "Black Friday"

category_compare = pd.concat([category_normal, category_bf])
```

```python
fig_category = px.bar(category_compare,x="product_category",y="purchase_amount",color="sale_type",barmode="group",title="Product Category Revenue Comparison", labels={"purchase_amount": "Revenue", "product_category": "Product Category"})

fig_category.update_layout(xaxis_tickangle=-45)

fig_category.show()
```

```python
fig_gender_donut.write_html("gender_chart.html")
fig_age.write_html("age_chart.html")
fig_category.write_html("category_chart.html")
```
