### HeatMap

```python
plt.figure(figsize=(8,6))

numeric_df = df.select_dtypes(include=np.number)

corr = numeric_df.corr()

mask = abs(corr) < 0.3

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    mask=mask,
    linewidths=0.5
)

plt.title("Important Correlations")

plt.show()
```
