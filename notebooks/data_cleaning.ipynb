## About dataset

### Name: Black Friday Retail Sales Analysis

The dataset represents Black Friday retail transactions collected from a retail business during festive sale periods.

Each row corresponds to a single customer transaction.

Primary objective:
1. Understand customer purchasing behavior
2. Identify profitable products and customer segments
3. Optimize marketing campaigns
4. Detect high-value customers

```python
# Importing libraries

import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("../datasets/retail_black_friday_sales_100k.csv")
df.head(5)
```

### Note: No NULL values and no Duplicate values detected in the dataset.

```python
# Type conversions

df['original_price'] = df['original_price'].astype('float64')
df['discount_pct'] = df['discount_pct'].astype('float64')
df['final_price'] = df['final_price'].astype('float64')
df['quantity'] = df['quantity'].astype('int64')
df['customer_id'] = df['customer_id'].astype('str')
df['transaction_id'] = df['transaction_id'].astype('str')
df['product_id'] = df['product_id'].astype('str')
df["purchase_date"] = pd.to_datetime(df["purchase_date"])

df.dtypes
```

```python
calculated_price = (df["original_price"] * (1 - df["discount_pct"] / 100)).round(2)

calculated_purchase = (df["final_price"] * df["quantity"]).round(2)
invalid_purchase = df[calculated_purchase != df["purchase_amount"]]
print(len(invalid_purchase))
```

```python
#Date metadata extraction
df["year"] = df["purchase_date"].dt.year
df["month"] = df["purchase_date"].dt.month
df["day"] = df["purchase_date"].dt.day
df["day_name"] = df["purchase_date"].dt.day_name()

def categorize_hour(hour):
    if 0 <= hour < 6:
        return "Late Night"

    if 6 <= hour < 12:
        return "Morning"

    if 12 <= hour < 18:
        return "Afternoon"

    return "Evening"


df["time_slot"] = df["purchase_hour"].apply(categorize_hour)
```

```python
df.head(6)
```

```python
customer_spending = (
    df.groupby("customer_id")["purchase_amount"]
    .sum()
    .reset_index()
)

customer_spending.rename(
    columns={
        "purchase_amount":
        "total_customer_spending"
    },
    inplace=True
)

customer_spending.head()
```

```python
Q1 = df["purchase_amount"].quantile(0.25)
Q3 = df["purchase_amount"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["purchase_amount"] < lower) |
    (df["purchase_amount"] > upper)
]

outliers.shape
```

## Data Analysis

### 1. customer purchase analysis

```python
#total revenue
print(f"Total Revenue: {df['purchase_amount'].sum()}") 

#Avg transaction value
print(f"Average Transaction Value: {df["purchase_amount"].mean()}")

print(f"Customer Count: {df["customer_id"].nunique()}")

top_customers = (df.groupby("customer_id")["purchase_amount"].sum().sort_values(ascending=False).head(10))
print(f"Top 10 Customers: \n{top_customers}")
```

### 2. Profitable product, age based and gender based spendings

```python
category_sales = (df.groupby("product_category")["purchase_amount"].sum().sort_values(ascending=False))

category_sales
```

```python
gender_sales = df.groupby("gender")["purchase_amount"].sum().sort_values(ascending=False)
gender_sales
```

```python
age_sales = df.groupby("age_group")["purchase_amount"].sum().sort_values(ascending=False)

age_sales
```

### 3. City sales and Segment-wise revenue

```python
city_sales = df.groupby("city")["purchase_amount"].sum().sort_values(ascending=False)
city_sales
```

```python
segment_sales = df.groupby("customer_segment")["purchase_amount"].sum()
segment_sales
```

### 4. High Value Customers

```python
customer_spending = df.groupby("customer_id")["purchase_amount"].sum().reset_index()

thresh = customer_spending["purchase_amount"].quantile(0.99)
high_value = customer_spending[customer_spending["purchase_amount"] > thresh]

high_value
```
